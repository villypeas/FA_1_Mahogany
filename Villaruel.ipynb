# Project Title: FA1_Distance-calculator

## Description
 It works by taking the values of the first point in the graph (x₁, y₁) and the second point (x₂, y₂) then it calculates the distance.

## How to Run
1. Enter the primary value (x₁, y₁)
2. Enter the secondary value (x₂, y₂)
3. Program will automatically calculate it and give you the answer

## Input Needed
- x₁, y₁, x₂ and y₂

## Sample Output
x₁ = 6, y₁ = 2, x₂ =5, y₂= 3

The answer is 1.4142135623730951
